```python
import yfinance as yf
import pandas as pd

ticker = "META"
stock = yf.Ticker(ticker)

financials = stock.financials
balance_sheet = stock.balance_sheet
cashflow = stock.cashflow
financials.head()

financials.to_csv("../data/raw/meta_financials.csv")
balance_sheet.to_csv("../data/raw/meta_balance_sheet.csv")
cashflow.to_csv("../data/raw/meta_cashflow.csv")

hist = stock.history(period="5y")
hist.to_csv("../data/raw/meta_prices.csv")

hist.head()
```
