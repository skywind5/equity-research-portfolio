```python
import pandas as pd
import matplotlib.pyplot as plt

financials = pd.read_csv("../data/raw/meta_financials.csv", index_col=0)

revenue = financials.loc["Total Revenue"]
net_income = financials.loc["Net Income"]

df = pd.DataFrame({
    "Revenue": revenue,
    "Net Income": net_income
}).T
df = df.T
df

df["Revenue Growth"] = df["Revenue"].pct_change()
df["Net Income Growth"] = df["Net Income"].pct_change()
df

plt.figure()
df["Revenue"].plot(title="Meta Revenue Trend")
plt.show()

plt.figure()
df["Net Income"].plot(title="Meta Net Income Trend")
plt.show()
```
